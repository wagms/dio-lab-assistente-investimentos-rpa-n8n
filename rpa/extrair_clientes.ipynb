# Código-fonte RPA (Python)

Extrai a lista de clientes da página hospedada no GitHub Pages e envia o resultado
para o Webhook do N8N (etapa de orquestração do workflow).

````python
import requests
from bs4 import BeautifulSoup

URL_PAGINA = "https://digitalinnovationone.github.io/dio-lab-assistente-investimentos-rpa-n8n/"

# 1) Baixar a página
html = requests.get(URL_PAGINA).text

# 2) Ler a tabela de clientes
soup = BeautifulSoup(html, "html.parser")
linhas = soup.select("#clientes tbody tr")

clientes = []

for linha in linhas:
    colunas = linha.find_all("td")

    cliente = {
        "nome": colunas[0].get_text(strip=True),
        "email": colunas[1].get_text(strip=True),
        "saldo": colunas[2].get_text(strip=True),
        "perfil": colunas[3].get_text(strip=True),
    }

    clientes.append(cliente)

# 3) Mostrar o resultado
print("Clientes encontrados:")
for cliente in clientes:
    print(cliente)

# 4) Enviar a lista "clientes" para o Webhook do n8n via POST.
#    Troque a URL abaixo pela URL do seu Webhook (aba "Production URL" ou
#    "Test URL" do node Webhook no seu workflow do n8n).
N8N_WEBHOOK = "https://SEU_N8N.app.n8n.cloud/webhook/clientes"

resposta = requests.post(N8N_WEBHOOK, json={"clientes": clientes})

print("\nStatus do envio ao N8N:", resposta.status_code)
print("Resposta do workflow (mensagens geradas):")
print(resposta.json())
````
